```python
# universe selection = 82,84,81,83,85
#chi-square selection = 85,87,86,84,88
#mutual info = 88,89,91,90,92
#using one way ANOVA test at 5% significance level:
#1. state the null hypothesis(H0) and alternate hypothesis(H1).
#2. calculate the mean accuracy for each feature-selection method.
#3. calculate the grand mean.
#4. calculate:between group sum of square(SSB) and within group sum of square(SSW)
#5. calculate the MSB,MSW
#6. calculate f statistic
#7. detarmine the critical f value at alpha = 0.05
#8. decide wheather H0 should be rejected or accepted
#9. interpreat the result in the context of ML feature selection

universe = [82, 84, 81, 83, 85]
chi_square = [85, 87, 86, 84, 88]
mutual_info = [88, 89, 91, 90, 92]
n = 5
mean_universe = sum(universe) / n
mean_chi = sum(chi_square) / n
mean_mi = sum(mutual_info) / n
print("Mean Accuracy:")
print("Universe Selection:", mean_universe)
print("Chi-square Selection:", mean_chi)
print("Mutual Information:", mean_mi)
all_data = universe + chi_square + mutual_info
grand_mean = sum(all_data) / len(all_data)
print("\nGrand Mean:", grand_mean)
SSB = n * (mean_universe - grand_mean) ** 2
SSB += n * (mean_chi - grand_mean) ** 2
SSB += n * (mean_mi - grand_mean) ** 2
print("\nSSB:", SSB)
SSW = sum((x - mean_universe) ** 2 for x in universe)
SSW += sum((x - mean_chi) ** 2 for x in chi_square)
SSW += sum((x - mean_mi) ** 2 for x in mutual_info)
print("SSW:", SSW)
k = 3
N = 15
df_between = k - 1
df_within = N - k
MSB = SSB / df_between
MSW = SSW / df_within
print("\nMSB:", MSB)
print("MSW:", MSW)
F = MSB / MSW
print("F statistic:", F)
F_critical = 3.885
print("Critical F value:", F_critical)
if F > F_critical:
    print("\nDecision: Reject H0")
    print("There is a significant difference between the methods.")
else:
    print("\nDecision: Accept H0")
    print("There is no significant difference between the methods.")
```

```python
#2. question
# Feature selection method: chi-square , ANOVA , mutual info , total
# High accuracy : 18,14,20,52
# Low accuracy : 7,11,5,23
# Total : 25,25,25,75
# using the chi-square test independence at 5% significance level:
#1. state H0 and H1
#2. constructed the expected frequency table
#3. calculate the chi-square statistic
#4. detarmine the degree of freedom
#5. find the critical chi-square value  at alpha = 0.05
#6. compute the calculate and critical value
#7. state your statistical decisison
#8. interpreate wheather feature selection method and model performance category are independend


# Chi-square test of independence

observed = [
    [18, 7],
    [14, 11],
    [20, 5]
]
row_totals = [sum(row) for row in observed]
col_totals = [
    sum(observed[i][j] for i in range(3))
    for j in range(2)]
grand_total = sum(row_totals)
expected = []
for i in range(3):
    row = []
    for j in range(2):
        E = (row_totals[i] * col_totals[j]) / grand_total
        row.append(E)
    expected.append(row)
print("Expected Frequency Table:")
for row in expected:
    print([round(x, 2) for x in row])
chi_square = 0
for i in range(3):
    for j in range(2):
        O = observed[i][j]
        E = expected[i][j]
        chi_square += (O - E) ** 2 / E
print("\nChi-square statistic:", round(chi_square, 2))
rows = 3
columns = 2
df = (rows - 1) * (columns - 1)
print("Degrees of freedom:", df)
critical_value = 5.991
print("Critical chi-square value:", critical_value)
if chi_square > critical_value:
    print("\nDecision: Reject H0")
    print("Feature selection method and performance are dependent.")
else:
    print("\nDecision: Fail to reject H0")
    print("Feature selection method and performance are independent.")
```

```python
#3. question
# student = 1,2,3,4,5,6,7,8
# study houe= 2,3,4,5,6,7,8,9
# ML marks = 45,50,54,60,65,70,74,80
# using the Karl pearson cor-relation coefficent
#1. identify the independend variable X and dependend variable Y
#2. detarmine the strength and direction of the cor-relation
#3. explain what the value of r indecates about study hours and ML examination marks
#4. does cor-relation imply that increased study hour neccessarily cause higher marks? explain

# Karl Pearson Correlation Coefficient

study_hours = [2, 3, 4, 5, 6, 7, 8, 9]
ml_marks = [45, 50, 54, 60, 65, 70, 74, 80]
import math
n = len(study_hours)
sum_x = sum(study_hours)
sum_y = sum(ml_marks)
sum_xy = sum(x * y for x, y in zip(study_hours, ml_marks))
sum_x2 = sum(x * x for x in study_hours)
sum_y2 = sum(y * y for y in ml_marks)
r = (n * sum_xy - sum_x * sum_y) / math.sqrt(
    (n * sum_x2 - sum_x ** 2) *
    (n * sum_y2 - sum_y ** 2))
print("Pearson correlation coefficient:", round(r, 3))
if r > 0:
    print("Direction: Positive")
else:
    print("Direction: Negative")
if abs(r) >= 0.9:
    print("Strength: Very Strong")
elif abs(r) >= 0.7:
    print("Strength: Strong")
elif abs(r) >= 0.5:
    print("Strength: Moderate")
else:
    print("Strength: Weak")
```
