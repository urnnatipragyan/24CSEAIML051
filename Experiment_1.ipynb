```python
1. # write a program to find the largest of three numbers

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))
if (a >= b and a >= c):
   print("Largest is:",a)
elif (b>c):
   print("Largest is:",b)
else:
  print("Largest is:",c)
```

```python
2. # write a program to check whether a number is even or odd.

num = int(input("Enter a number: "))
if (num % 2 == 0):
   print("The number is even")
else:
   print("The number is odd")
```

```python
3. # write a program to swap 2 numbers without using 3rd variable

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
print("Before swapping:a=",a,"b=",b)
a = a+b
b = a-b
a = a-b
print("After swapping:a=",a,"b=",b)
```

```python
4. # write a program to check wheather a string is palindrome

str = input("Enter a string: ")
if (str == str[::-1]):
   print("The string is palindrome")
else:
   print("The string is not palindrome")
```

```python
5. #write a program to implement a simple calculator

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter an operator (+,-,*,/): ")
if(op == "+"):
   print(a+b)
elif(op == "-"):
   print(a-b)
elif(op == "*"):
   print(a*b)
elif(op == "/"):
   print(a/b)
else:
  print("please enter a valid input")
```

```python
6. # write a program to count vowels in a string

str = input("Enter a string: ")
count = 0
for i in str:
  if i in "aeiouAEIOU":
    print("vowel")
    count = count+1
  else:
    print("No vowel")
print("Total number of vowel are:",count)
```

```python
7. # write a program to check wheather a number is prime

num = int(input("Enter a number: "))
count = 0
for i in range(1,num+1):
  if (num%i == 0):
     count+=1
if count == 2:
  print("The number is prime")
else:
  print("The number is not prime")
```

```python
8. # write a program to find the sum of digits of number

num = int(input("Enter the numbers: "))
sum = 0
while num>0:
  rem = num%10
  sum = sum+rem
  num = num//10
print("The sum of digits of number is:",sum)
```

```python
9. # write a program to reverse a number

n = int(input("Enter the number: "))
rev = 0
while n>0:
  digit = n%10
  rev = rev*10+digit
  n//=10
print("The reverse of number is:",rev)
```

```python
10. # write a program to find the factorial of a number

n = int(input("Enter the number: "))
fact = 1
for i in range(1,n+1):
   fact*=i
print("The factorial of number is:",fact)
```
