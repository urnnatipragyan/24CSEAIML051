```python
import numpy as np
import matplotlib.pyplot as plt

def predict(x,w,b):
  return w*x + b

def compute_cost(y, y_pred, w, lambda_reg, n):
  mse_cost = np.sum((y-y_pred)**2)/(2*n)
  l1_regularization = lambda_reg * np.abs(w)
  total_cost = mse_cost + l1_regularization
  return total_cost

def gradient(x,y,y_pred, w, lambda_reg, n):
  dw = (np.sum(x * (y_pred - y))) / n
  db = (np.sum(y_pred - y)) / n
  # Add L1 regularization subgradient for dw
  dw += lambda_reg * np.sign(w)
  return dw, db

def parameter_update(w,b,dw,db,alpha):
  w = w - alpha*dw
  b = b - alpha*db
  return w,b

x = np.array([1,2,3], dtype=float)
y = np.array([2,3,4], dtype=float)

n = len(x)
print("Input x : ", x)
print("Actual y :", y)
print("Number of samples : ", n)

w = 0.0
b = 0.0
alpha = 0.1
itterations = 10 # Changed iterations to 10 as requested
lambda_reg = 0.1 # Lasso regularization parameter
print("Initial weight (w) : ", w)
print("Initial bias (b) : ", b)
print("Learning rate : Adjust to 0.01 for better convergence", alpha)
print("Lasso regularization (lambda_reg) : ", lambda_reg)

y_pred = predict(x,w,b)
print("Predictions : ", y_pred)


cost = compute_cost(y,y_pred, w, lambda_reg, n)
print("\nInitial prediction : ", y_pred)
print(f"Initial Cost : {cost:.4f}")

costs = []
w_history = []
b_history = []


for i in range(itterations):
  y_pred_current_params = predict(x,w,b)
  dw, db = gradient(x,y,y_pred_current_params, w, lambda_reg, n)
  w, b = parameter_update(w,b,dw,db,alpha)
  y_pred_after_update = predict(x,w,b)
  cost_after_update = compute_cost(y,y_pred_after_update, w, lambda_reg, n)

  costs.append(cost_after_update)
  w_history.append(w)
  b_history.append(b)
  # Print details for every iteration
  print("\n"+"-"*30)
  print(f"Iteration {i+1}")
  print(f"The y_predict (after update) is {np.round(y_pred_after_update, 4)}")
  print(f"The cost (after update) is {cost_after_update:.4f}")
  print(f"Gradients dw: {dw:.4f}, db: {db:.4f}")
  print(f"The updated weight (w) is {w:.4f}")
  print(f"The updated bias (b) is {b:.4f}")


final_w = w_history[-1]
final_b = b_history[-1]
print("\nThe final equation is : ")
print(f"y = {final_w:.4f}x + {final_b:.4f}")
print("The final cost is ", costs[-1])


```
