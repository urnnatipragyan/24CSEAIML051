```python
# Let us consider the dataset which consists of 6 observation with 2 features and a target
# obs= 1,2,3,4,5,6
# Rain fall = high,high,high,low,low,low
# soiltype =  Loamy,sandy,clay,Loamy,sandy,clay
# yield = good,good,good,poor,poor,good
# use oneR to determine which feature give the best single rule classify

rainfall = ["high", "high", "high", "low", "low", "low"]
soiltype = ["loamy", "sandy", "clay", "loamy", "sandy", "clay"]
yield_value = ["good", "good", "good", "poor", "poor", "good"]
features = {
    "Rainfall": rainfall,
    "SoilType": soiltype
}
best_feature = None
best_error = 999
feature_errors = {}
for feature_name, feature_values in features.items():
    print("\nFeature:", feature_name)
    total_error = 0
    for value in set(feature_values):
        target_values = []
        for i in range(len(feature_values)):
            if feature_values[i] == value:
                target_values.append(yield_value[i])
        good = target_values.count("good")
        poor = target_values.count("poor")
        if good >= poor:
            prediction = "good"
            error = poor
        else:
            prediction = "poor"
            error = good
        print("IF", feature_name, "=", value,
              "THEN Yield =", prediction)
        total_error = total_error + error
    feature_errors[feature_name] = total_error
    error_percentage = (total_error / len(yield_value)) * 100
    print("Total Error =", total_error)
    print("Error Percentage =", round(error_percentage, 2), "%")
    if total_error < best_error:
        best_error = total_error
        best_feature = feature_name
print("\n----------------------------")
print("Feature Ranking")
print("-----------------------------")
rainfall_error = feature_errors["Rainfall"]
soil_error = feature_errors["SoilType"]
print("1. Rainfall  =", round((rainfall_error / 6) * 100, 2), "%")
print("2. SoilType  =", round((soil_error / 6) * 100, 2), "%")
print("\nSelected Feature:", best_feature)
print("Minimum Error =", round((best_error / 6) * 100, 2), "%")
```
